#Dictionary:
are used to store data values in "key":"values" pair.

They are unorderd, mutable but dose not allow duplicate keys.

```python
#syntax
dictonary = {

    "occupation": "student",
    "key2": 3,
    "key3": 4.5,
    "key4": [23,456,76],
    "key5":("alex","bob",'casey')

}


print(dictonary)
print(type(dictonary))
print(len(dictonary))
```

Create a dictionary to store student name, id, cgpa, list of subjects name, subjects code as tuples & take student_info as dictionary name to create a dictionary

```python
student_info = {
    "name" : "Alex",
    'id' : 23454,
    'cgpa': 3.5,
    "subjects":['phy','chem','bio'],
    'sub_code':(101, 102,103)

}
#print the entire dictionary
print(student_info)


#Print name of student & subjects
print(student_info['name'])
print(student_info['subjects'])
print(student_info["sub_code"])
```

```python
#chage name to Bob
student_info ['name'] = "Bob"
print(student_info)
```

```python
#add last name as new values
student_info['last_name'] = 'Smith'
print(student_info)
```

```python
#empty / null dictionary
student21 = {}
student22 = {}
student23 = {}

#print a null dictionary
print(student21)
print(student22)
print(student23)

#store name = "casey" and age = 23
student21['name'] ='casey'
student21['age'] = 23

print(student21)
print(student22)
print(student23)
```

```python
#Nested dictionary
#created a student dictionary store name, subject and make it a nested dictionary to store subjects and their marks
student = {
    'name': 'Donald',
    "subject" :{
        'phy': 101,
        'chem':102,
        'math': 103
    },

    'marks':{
        'phy': 90,
        'chem': 80,
        'math': 70
    }
}


#print the entire dictionary
print(student)

#print only subjects
print(student['subject'])

#print math marks
print(student['marks'] ['math'])
```

#Methods

```python
#.keys()- return all keys
print(student)
print(student.keys())
print(list(student))#type cast
print(len(list(student.keys())))
```

```python
# .values() - return all values
print(student.values())
```

```python
# .items() - return all ("keys": "values") pairs as tuples
print(student.items())

#Type cast into list
print(list(student.items()))

#2nd way(dictionary directy can not access index number, to access index number have to typecast)
stu = list(student.items())
print(stu[0])

#print index 0
print(list(student.items())[0])
```

```python
# .get("keys_name") - return all the keys according to values
print(student.get('name'))#is name dosent exist return none

#if name doesnt exist show error
print(student['name'])
```

```python
# .update() -insert the specific items to the dictionary
student.update({
    'city':"Dhaka",
    'age':25,
    'college': "DMC"})
print(student)
```
