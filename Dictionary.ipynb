#Dictionary:
are used to store data values in "key":"values" pair.

They are unorderd, mutable but dose not allow duplicate keys.

```python
#syntax

Dictionary ={
    "keys":"vales",
    "key1":"str",
    "key2":"int",
    "key3":"float",
    "key4":"list",
    "key5":"tuples"
}

print(Dictionary)
print(type(Dictionary))
print(f"Length is: {(len(Dictionary ))}")
```

Create a dictionary to store student name, id, cgpa, list of subjects name, subjects code as tuples & take student_info as dictionary name create a dictionary

```python
student_info = {
    "name" : "Alex",
    "id" : 23456,
    "cgpa" : 3.45,
    "Sub_name" : ["DSA", "EEE", 'Physics'],
    "Sub_code" : (101, 205, 111)
}
#print the entire dictionary

print(student_info)

#Print name of student & subjects

print(student_info["name"])
print(student_info["Sub_name"])
```

```python
#chage name to Bob
student_info["name"] = 'Bob'
print(student_info)
```

```python
#add last name as new values
student_info["Last_name"] = "Bahar"
print(student_info)
```

```python
#empty / null dictionary
null_info = {}
#print a null dictionary
print(null_info)

#store name = "casey" and age = 23
null_info['name'] = 'casey'
null_info ['age'] = 23
print(null_info)
```

```python
#Nested dictionary
#created a student dictionary store name, subject and make it a nested dictionary to store subjects and their marks
student ={
    "name" : "Donald",
    'Subject' : {
        "phy":34,
        'chem': 67,
        'math' : 99

    }
}
#print the entire dictionary
print(student)

#print only subjects
print(student['Subject'])

#print math marks
print(student["Subject"]["math"])
```

#Methods

```python
#.keys()- return all keys
print(student.keys())

#change keys in the formate of list (type cast)
print(list(student.keys()))

#print the length of that list
print(len(list(student.keys())))

print(len(student.keys()))
```

```python
# .values() - return all values
print(student.values())
```

```python
# .items() - return all ("keys": "values") pairs as tuples
print(student.items())

#Type cast into list
pair = list(student.items())
print(pair)

#print index 0
print(pair[0])
```

```python
# .get("keys_name") - return all the keys according to values
print(student.get("name"))

print(student['name'])
```

```python
# .update() -insert the specific items to the dictionary

#insert city and age
student.update({"city":'Dhaka', "age" : 23})

print(student)
```
