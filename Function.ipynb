```python
#print factiorial of number n
#if n=5! = 1*2*3*4*5

n = int(input("Enter a number: "))
def cal_fact(n):
    fact = 1
    for i in range(1, n+1):#for is used to calculate index i
        fact = fact * i
    print(fact)
    return fact
x=cal_fact(n)
```
