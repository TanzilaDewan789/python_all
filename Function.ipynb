```python
#return- sends a value back from a function.

def add(a,b):
  s = a+b
  print(s)
  return s #(return s sends the value of s back to wherever the function was called.)

x = add(20,89)
print(x)
```

```python
#print factiorial of number n
#if n=5! = 1*2*3*4*5
n = int(input("Enter a number: "))
def cal_fact(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact * i
    print(fact)
    return fact
x=cal_fact(n)
```
