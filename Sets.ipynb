#Sets
Set is a collection of unorderd items. Each element in the set is unique & immutable.
It store - boolean, int, float, str, tuple.

can not store list and dictionary.

set element - immutable

set - mutable



```python
collection = {1,2,3,4,4,4,4,4, 9.8, 'hello', 'hello','world',(1,2,3,4),(1,2,3,4)}
#print-unorderd and remove duplicates values
print(collection)
print(type(collection))

print(len(collection))#length counted wkithout duplicate value
```

```python

#print an empty set
name = set()
```

#Method

```python
#.add(element) - add element to set
name.add('Adam')
name.add('Bob')
name.add('Casey')
name.add(1)
name.add(1)
name.add(3.4)
name.add(("Donald","Emli","Frank"))

print(name)
```

```python
#.remove(element)
name.remove('Bob')
print(name)
```

```python
#.pop - remove a random value
name.pop()
print(name)
```

```python
#.clear - return an empty set
name.clear()
print(name)
```

# Union and Intersection


```python
set1 = {1,2,3,4,5,6}
set2 = {3,4,5,6,7,8,9}

#.union - combined both and remove duplicates and return all
print(set1.union(set2))

#.intersection - combined both return common values
print(set1.intersection(set2))
```

```python
# and(&) or(|)

#and(&) as same intersection
set1 = {1,2,3,4,5,6}
set2 = {3,4,5,6,7,8,9}
print(set1 & set2)

#or(|) as same union
print(set1 | set2)
```
