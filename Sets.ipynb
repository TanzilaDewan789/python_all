#Sets
Set is a collection of unorderd items. Each element in the set is unique & immutable.
It store - boolean, int, float, str, tuple.

set element - immutable

set - mutable



```python
collection = {1,2,3,3.4,"hello", 2,5,'hello', (4,5,6)}
print(collection) #print-unorderd
print( type( collection))
print(len(collection))# remove duplicates values
```

```python
#print an empty set
collection2 = set()
print(collection2)
```

#Method

```python
#.add(element) - add element to set
collection2.add(1)
collection2.add(1)
collection2.add(2)
collection2.add(2)
collection2.add(3)
collection2.add(4)
collection2.add('adam')
collection2.add((1,2,3,4,5))
print(collection2)
```

```python
#.remove(element) - remove an element
collection2.remove(2)
print(collection2)
```

```python
#.pop - remove a random value
print(collection2.pop())

```

```python
#.clear - return an empty set
collection2.clear()
print(collection2)
```

# Union and Intersection


```python
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8,9,10}

#.union - combined both and remove duplicates and return all
print(set1.union(set2))

#.intersection - combined both return common values
print(set1.intersection(set2))
```
