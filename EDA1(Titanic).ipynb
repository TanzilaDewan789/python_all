#Uni_variate Analysis
(one variable) or 1 column analysis
1. numerical analysis (int or float)
2. catagorical analysis (object)

```python
#catagorica analysis
#for Survived class
sns.countplot(data =df,
              x = 'Survived')
```

```python
survived_count = df['Survived'].value_counts()
survived_count
```

```python
# rate of survived people
df['Survived'].value_counts()/len(df)*100
```

```python
survived_label = df['Survived'].unique()
#print(survived_label)
plt.pie(survived_count, autopct ='%1.1f', labels= survived_label)

plt.title("Pie chart of survival rate")
plt.show()
```

```python
#catagorica analysis
# for Pclass
sns.countplot(data =df,
              x = 'Pclass')
```

```python
(df['Pclass']).value_counts()/len(df)*100
```

```python
#Catagory analysis - for gender
sns.countplot(data= df,
              x = 'Sex')
```

```python
(df['Sex']).value_counts()/len(df)*100
```

```python
#Catagorical analysis - embark
sns.countplot(data= df,
              x = 'Embarked')
```

```python
(df['Embarked']).value_counts()/len(df)*100
```

#Uni-variate Analysis

```python
#Numerical analysis for age
sns.histplot(df['Age'], bins = 50)

plt.title('Histogram of Age')
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('KDE plot of Age')
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
#Numerical analysis for for fare
sns.histplot(df['Fare'], bins=50)

plt.title('Histogram of Fare')
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title('KDE plot of Fare')
plt.xlabel("Fare")
plt.ylabel("Probability distributions")

plt.show()
```

```python
#Boxplot for outliers
sns.boxplot(df['Fare'])
plt.title('Boxplot of Fare')
plt.xlabel("Fare")
```

#Multi- variate analysis

working on multiple columns is known as multi - variate analysis.

If only work on 2 coulumns is known as bi - variate analysis.

```python
# catagorical( sex ) and numerical( survived ) column relation
sns.countplot(data = df,
              x = 'Sex',
              hue = 'Survived')
```

```python
gender_split = df.groupby('Sex')['Survived'].value_counts(normalize=True)
#femal yes = (233/233+82)*100
gender_split
```

```python
sns.countplot( data = df,
              x = 'Pclass',
               hue= 'Survived')
```

```python
pclass_split = df.groupby('Pclass')['Survived'].value_counts(normalize=True)
pclass_split
```

```python
sns.barplot(data=df,
            x='Pclass',
            y='Fare')
```

```python
sns.barplot(data = df,
            x ='Survived',
            y = 'Age',
            hue = 'Sex')
```
