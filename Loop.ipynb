#While loop-
Used when you want the loop to continue until a condition becomes false.

Syntax-

```python
#print number 1 to 5

i = 1
while i <=5:
  print(i)
  i+=1
```

```python
#print odd number from 1 to 10
i = 1
while i <=10:
  print(i)
  i+=2
```

```python

#print number 10 to 1
i = 10

while i>=1:
  print(i)
  i-=1

```

```python
#print even numbers backward from 10 to 1
i =10
while i >= 1:
  print(i)
  i-=2
```

```python
#print multiplication table of a number n
n = int(input("Enter a number for multiplication table: "))
i = 1
while i <= 10 :
  print(f"{n}*{i} = {n*i}")
  i+=1
```

```python
#Print the element of following list [1,4,9,16,25,36,49,64,81,100]

nums = [1,4,9,16,25,36,49,64,81,100]
i = 0
while i < len(nums):
  print(nums[i])
  i+=1
```

#Break & Continue
break and continue are used to control loops.

```python
#break: stops the loop immediately.
i = 1
while i <=10:
  print(i)
  if(i==3):
    break
  i +=1
```

```python
# Traverse:
# find x in this tuple

num = (1,4,9,16,25,36,49,64,81,100)
x = 36

i = 0

while i < len(num):
  if(num[i]==x):
    print(f'Found at index {i}')
    break

  else:
    print("Finding...")
    i += 1
```

```python
#continue: skips the current iteration and moves to the next loop cycle.
i = 0
while i <= 5:
  if( i==3):
    i += 1
    continue #skip one step
  print(i)
  i += 1
```

```python
#print odd number from 1 to 20 using while loop with continue
i = 1

while i<=20:
  if (i%2 == 0):
    i+=1
    continue
  print(i)
  i += 1

```

```python
#print even numbers from 0 to 20

i = 0

while i<=20:
  if (i%2 != 0):
    i+=1
    continue
  print(i)
  i += 1
```
