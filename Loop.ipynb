#Loop
Use to repeat instructions. 1. for loop 2. while loop

#While loop-
Used when you want the loop to continue until a condition becomes false.

Syntax-

initialization

while condition:

print('statment')

update-increment or decrement

```python
#print number 1 to 5
i = 1
while i <= 5:
  print(i)
  i += 1
```

```python
#print odd number from 0 to 10
i = 1
while i <= 10:
  print(i)
  i += 2

```

```python

#print number 10 to 1
i = 10
while i >= 1:
  print(i)
  i -= 1

```

```python
#print even numbers backward from 10 to 1
i = 10
while i >= 1:
  print(i)
  i -= 2
```

```python
#print multiplication table of a number n
n = int(input("Enter any number "))
i = 1
while i <= 10:
  print(f'{n}*{i}={n*i}')
  i += 1
```

```python
#Print the element of following list [1,4,9,16,25,36,49,64,81,100]
n =  [1,4,9,16,25,36,49,64,81,100]
i = 0 #index (0 to 9)

while i < len(n):#len(n)= 10
  print(n[i])
  i += 1

```

#Break & Continue
break and continue are used to control loops.

```python
#break: stops the loop immediately.
i = 1
while i <= 10:
  print(i)
  if(i == 7):
    break
  i += 1


```

```python
# Traverse: Linear search
# find x = 49 in this tuple

n = (1,4,9,16,25,36,49,64,81,100) #len-10
i = 0
while i < len(n):
  if(n[i]==49):
    print(f"49 Found at index {i}")
    break
  else:
    print("Finding...")
    i+= 1
```

```python

#continue: skips the current iteration and moves to the next loop cycle.
i = 1
while i <= 10:
  if(i == 4):
    i += 1
    continue

  print(i)
  i += 1
```

```python
#print odd number from 1 to 10 using while loop with continue
i = 1
while i <= 10:
  if(i%2 == 0):#even
    i += 1 #even + 1 = odd
    continue

  print(i)
  i += 1
```

```python
#print even numbers from 0 to 20
i = 0
while i <= 20:
  if(i%2 != 0):
    i += 1
    continue

  print(i)
  i += 1

```

#For loop
Use to iterate over sequences such as list, tuples, strings and range.

```python
num = (1,2,3,4,5)
for i in num:
  print(i)
```

```python
#find 'o' in "Python" if you get print "found" until print all the character and if there is no 'o' print "not found"
n = 'Python'
for i in n:
  if(i == "o"):
    print('o Found')
    break
  print(i)
else:
  print("Not found")

```

```python
# "cat" find index number for char "t"
animal = "Cat"
idx = 0
for i in animal:
  if(i == "t"):
    print(f't found at index {idx}')
    break
  idx += 1
  print(i)
```

```python
#enumerate as index number

animal = "Puppy"

for j,i in enumerate(animal):
  if(i =='y'):
    print(f'y found at index {j}')
    break
  print(i)

else:
  print("Not found")
```

```python
n = [1,4,9,16,25,36,49,64,81,100]
for j,i in enumerate(n):
  if(i == 49):
    print(f'49 found at index {j}')
    break
  print(i)
else:
  print("not found")


```

```python
#range is a built in function that return a sequence of numbers.By default starting from 0, increment 1, and stop before a specif number.
for i in range(10):
  print(i)
```

```python
#start, stop
for i in range (1,11):
  print(i)
```

```python
#start,stop,step
for i in range(1,11,1):
  print(i)
```

```python
#print even number from 1 to 10
for i in range(2,11,2):
  print(i)
```

```python
#print 10 to 1
for i in range(10,0,-1):
  print(i)
```

```python
#print multiplication table of a number n
n = int(input("Enter any number"))

for i in range(1,11,1):
  print(i*n)
```

```python
#Find the sum of first natural number n=5 (for & while loop)
#n = 5(1+2+3+4+5=15)

#for
n = 5
total = 0
for i in range(1,n+1):
  total = total + i
print(total)
```

```python
#while
n = 5
t = 0
i = 1
while i <= n:
  t = t + i
  i += 1
  print(t)
```

```python
#Factorial n =5! = 1*2*3*4*5 =120
#for
n = 5
fact =1
for i in range(1,n+1):
  fact = fact * i
  print(fact)
```

```python
#While
n =5
i =1
fact = 1
while i <= 5:
  fact = fact * i
  i += 1
print(fact)
```
