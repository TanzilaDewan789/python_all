#1.Inheritance
When one class inherits some features from another class this phenomena is known as inheritance.

Example:SportsCar inherits features from Car.

Benefit:Code reusability and reduced duplication.

```python
class Parents: #Parent class/ Super class
  def speak(self):
    print("I can speak!")

class Children(Parents): #Child class/ sub class
  pass

#object for parent class
obj = Parents()
obj.speak()

#Object for child class
obj1= Children()
obj1.speak()
```

```python
#Constructor in Inheritance

class Animal:#parent
  def __init__(self,name):#constructor
    self.name = name

  def show(self):
    print(f"I am {self.name}")

class Human(Animal):#child
  pass

#Objects
animal1 = Animal("Lion")
animal1.show()

human1 = Human("Adam")
human1.show()
```

#super() function

```python
class Animal:
  def __init__(self, name):
    self.name = name

  def show(self):
    print(f"Hello, my name is: {self.name}, I dont know my age")

class Human(Animal):
  def __init__(self, name, age):
      super().__init__(name)
      self.age= age

  def show(self):
    print(f"My name is {self.name} and I am {self.age} years old")


animals1 = Animal("Lion")
animals1.show()

person1 = Human("Bob", 23)
person1.show()
```

#5 types of Inheritance

#1.Single Inheritance
One parent → One child


```python
class Animal:
    def parrots(self):
        print("I can speak!")

class Bird(Animal):
    pass



obj = Animal() # Create object (instance) of parent class
obj.parrots() # Call method of parent class

obj2 = Bird() # Create object (instance) of child class
obj2.parrots()# Call inherited method using child object
```

#2. Multiple Inheritance
Multiple parents → One child

```python
class Animal:
  name1 = "Lion"

class Human:
  name2 = "Casey"

class Robot(Animal, Human):
  name3 = "WALL-E"

obj = Robot()
print(obj.name1)
print(obj.name2)
print(obj.name3)
```

```python
#MRO(Method Resolution Order)-Inherites first class
class Animal:
  def __init__(self,name):
    pass

class Human:
  def __init__(self,name,age):
    pass

class Robot(Human,Animal):#Human class inherites first
  def __init__(self,name,age):
    super().__init__(name,age)
    self.name = name
    self.age = age

obj = Robot("WALL-E",2)
print(obj.name)
print(obj.age)

print(Robot.mro())
```

#3.Multilevel
Grandparent → Parent → Child

```python
class Main_Factory:#Grandparent
  def __init__(self,materials,zips):
    self.materials = materials
    self.zips = zips

class Factory_2nd(Main_Factory):#Parent
    def __init__(self,materials,zips,colors):
      super().__init__(materials, zips)
      self.colors = colors

class Factory_3rd(Factory_2nd):#Child
    def __init__(self,materials, zips, colors, pockets):
      super().__init__(materials,zips,colors)
      self.pockets = pockets


obj = Factory_3rd("Leather", 2, "Black", 4)

print(obj.materials)
print(obj.zips)
print(obj.colors)
print(obj.pockets)
```

#4.Hierarchical
One parent → Multiple children

```python
class Animal:
    def speak(self):
        print("Speaking")

class Dog(Animal):
    pass

class Cat(Animal):
    pass

d = Dog()
c = Cat()

d.speak()
c.speak()

"""
      Animal
      /    \
    Dog    Cat


    """
```

#5.Hybrid
Combination of inheritance types

```python
class A:
    def show(self):
        print("Class A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

d = D()
d.show()

"""   A
     / \
    B   C
     \ /
      D
      """
```
