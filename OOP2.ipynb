#1.Inheritance
When one class inherits some features from another class this phenomena is known as inheritance.

Example:SportsCar inherits features from Car.

Benefit:Code reusability and reduced duplication.

```python
class Parents: #Parent class/ Super class
  def speak(self):
    print("I can speak!")

class Children(Parents): #Child class/ sub class
  pass

#object for parent class
obj = Parents()
obj.speak()

#Object for child class
obj1= Children()
obj1.speak()
```

```python
#Constructor in Inheritance

class Animal:#parent
  def __init__(self,name):#constructor
    self.name = name

  def show(self):
    print(f"I am {self.name}")

class Human(Animal):#child
  pass

#Objects
animal1 = Animal("Lion")
animal1.show()

human1 = Human("Adam")
human1.show()
```
