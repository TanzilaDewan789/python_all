#Series
Serias is one dimansion labeled array capable of holding any data type.

Axis labels are collectively called the index.(Index=Label)

Series = One Column

```python
#Import library
import numpy as np
import pandas as pd
```

```python
#Creating series
alphabet=["a","b","c"] #one dimension(1D)
digit =[1,2,3]

num=[[10,20,30],
     [60,70,80]] #2D-like table(matrix)

arr = np.array([10,20,30])
d = {
    'school':"ABC school",
    "name":"Bob",
    "age":12

}

#printing series 1D
pd.Series(alphabet) #1D = Line of values
```

```python
#make alphabet as index and digit as series
pd.Series(digit, index= alphabet)
```

```python
#print 2D
pd.Series(num)
```

```python
#pritn array
pd.Series(arr)
```

```python
#print dictionary
pd.Series(d)
```

```python
#3D - stack of table(stack of matrix)
arr = np.array([
    [[0, -3], [5, 2]],
    [[13, 1], [-4, 5]]
])

s = pd.Series(list(arr))#No native 3D structure in Pandas

print(s)
print(arr.ndim)# show dimention number -3
print(arr.shape)#number of- matrices, rows, columns(2,2,2)

"""
Matrix-A
[ 0  -3 ]
[ 5   2 ]

Matrix-B
[ 13   1 ]
[ -4   5 ]

Output-
[
    Matrix A,
    Matrix B
]

2 matrices(A,B)
2 rows per matrix
2 columns per row
"""
```

#DataFrame
DataFrame = Full Table

A DataFrame is a two-dimensional table made up of rows and columns.

Entire table → DataFrame

Name column only → Series

Salary column only → Series

```python
#creating a table using DataFrames
#dictionary(key-heading:value-table data)
data = {
    'name': ['John','Bob','Arica','Anna'],
    'Age':[28,34,29,42],
    'city':['New Yourk','Paris','Berlin','London'],
    'salary':[65000,80000,60000,90000]
}
#print DataFrame
pd.DataFrame(data)
```

```python
#creating data frame as list without the heading
data_list=[
    ['John', 28, 'New York', 65000],
    ['Bob', 34, 'Paris', 80000],
    ['Arica', 29, 'Berlin', 60000],
    ['Anna', 42, 'London', 90000]

]
df = pd.DataFrame(data_list)
print(df)#Storing for later
```

```python
#adding column name to the table
col = ['Name','Age','City','Salary']
df = pd.DataFrame(data_list,columns = col)
df
```

```python
#Add new row
df.loc[len(df)] = ['Donald', 33, 'Dubai', 45000]
df
```

```python
#Accessing / selecting index of column
#show name
df['Name']
```

```python
#Select multiple column
#show name & city
df[['Name','City']]
```

```python
#creating a new column
df['Designation']=['Engg.','Doctor','Teacher','Bus Driver','Manager']
df
```

```python
#delete a column/ row (temp.)
df.drop('Designation',axis= 1) #axis= 1 for deleting column/ X axis
#df.drop(0, axis= 0)            #axis= 0 for deleting column/ Y axis
```

```python
df #Designation is still available
```

```python
#Delete column permanently
df.drop('Designation', axis = 1, inplace= True)
df
```

```python
#delete multiple column(temp.)
df.drop(['Name','City'], axis=1)
```

```python
#Selecting Rows- 1st 2 rows
df.loc[[0,1]]
```

```python
#Index location
df.iloc[3]
```

```python
#Selecting subset of rows/column
df.loc[[0,1],['City','Salary']]
```

```python
#Conditional selection
#Show those whose age is more then 30
df[df['Age']>30]
```

```python
#age > 30 and live in Paris
df[(df["Age"]>30) & (df['City']=='Paris')]
```

```python
#age > 30 or live in paris
df[(df['Age']>30) | (df['City']=='Paris')]
```
