#Margin_Joining_Concatination

#Margin

```python
#Creating 2 dictionary table to marge 2 data frame
employees =pd.DataFrame({
    "emp_id":[1,2,3,4,5],
    "name":['John','Anna','Peter','Linda','Bob'],
    "dept":['HR','IT','Finance','IT','HR']
})

salaries = pd.DataFrame({
    "emp_id":[1,2,3,6,7],
    "salary":[60000,80000,650000,70000,90000],
    "bonus":[5000,1000,7000,8000,12000]
})
```

```python
#showing 1st table
employees
```

```python
#showing 2nd table
salaries
```

```python
#Merge 2 table
pd.merge(employees,salaries)
```

```python
#on- work as join key or common column
pd.merge(employees,salaries, on = "emp_id") #on = common column in both table
```

```python
 #returning only the rows with matching emp_id values in both DataFrames.
pd.merge(employees,salaries, on = "emp_id", how = "inner")
```

```python
#returning all rows from both DataFrames, filling non-matching values with NaN.
pd.merge(employees,salaries, on = "emp_id", how = "outer")
```

#GroupBy

```python
data = {
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['s1','s1','s2','s2','s1','s2','s2','s1'],
    'Sales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01', periods =8)
}

table = pd.DataFrame(data)
table
```

```python
#Grpup by catagory A & B
cat = table.groupby('Category')

for i, v in cat:
  print(i)#i = the group name (key) ('A' or 'B')
  print(v)#v = the DataFrame containing all rows belonging to that group

cat #group by create an object that can't see group A & B directly
```

```python
#another way for A
cat = table.groupby('Category').get_group("A")
cat
```

```python
# another way for B
cat = table.groupby("Category")
cat.get_group("B")
```

```python
#sum of sales
cat = table.groupby("Category")["Sales"].sum()
cat
```

```python
#group by store and calculate the sum of sales
cat = table.groupby("Store")['Sales'].sum()
cat
```

```python
#Group by multiple column(category and store) and calculate the sum of sales
cat = table.groupby(['Category',"Store"])['Sales'].sum()
cat
```

#Aggregation
mean, median, mode, min, max, count,std

```python
#calculates the average(mean)
table['Sales'].mean()
```

```python
#can perform multiple calculation at once
table["Sales"].agg(['sum','mean','min','max','count','std','median'])
```

```python
#multiple group to perform multiple calculation
table.groupby('Category')['Sales'].agg(
    ['sum','mean','min','max','count']
)
```

#Pivot Tables
A Pivot Table is a tool that summarizes and analyzes data by grouping it into categories.

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),#(lowest,highest, 20 random numbers)
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

pt = pd.DataFrame(data)
pt
```

```python
#creates a Pivot Table that shows the median sales for each combination of Region and Product.
pd.pivot_table(pt,
    values = 'Sales',
    index = 'Region',
    columns = 'Product',
    aggfunc = 'median')
```

```python
#creates a pivot table using two value columns: Sales and Units.
pd.pivot_table(
    pt,
    values=['Sales','Units'],
    index = 'Region',
    columns = 'Product'
    )
```

#Cross Table
a Cross Table (Crosstab) is a table that counts how many times two categories occur together.

```python
#Displays the frequency (count) of each Product for every Region.
pd.crosstab(pt['Region'],pt['Product'])#[index],[column]
```
