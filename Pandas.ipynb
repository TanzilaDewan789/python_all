#Series
Serias is one dimansion labeled array capable of holding any data type.

Axis labels are collectively called the index.(Index=Label)

Series = One Column

```python
#Import library
import numpy as np
import pandas as pd
```

```python
#Creating series
alphabet=["a","b","c"] #one dimension(1D)
digit =[1,2,3]

num=[[10,20,30],
     [60,70,80]] #2D-like table(matrix)

arr = np.array([10,20,30])
d = {
    'school':"ABC school",
    "name":"Bob",
    "age":12

}

#printing series 1D
pd.Series(alphabet) #1D = Line of values
```

```python
#make alphabet as index and digit as series
pd.Series(digit, index= alphabet)
```

```python
#print 2D
pd.Series(num)
```

```python
#pritn array
pd.Series(arr)
```

```python
#print dictionary
pd.Series(d)
```

```python
#3D - stack of table(stack of matrix)
arr = np.array([
    [[0, -3], [5, 2]],
    [[13, 1], [-4, 5]]
])

s = pd.Series(list(arr))#No native 3D structure in Pandas

print(s)
print(arr.ndim)# show dimention number -3
print(arr.shape)#number of- matrices, rows, columns(2,2,2)

"""
Matrix-A
[ 0  -3 ]
[ 5   2 ]

Matrix-B
[ 13   1 ]
[ -4   5 ]

Output-
[
    Matrix A,
    Matrix B
]

2 matrices(A,B)
2 rows per matrix
2 columns per row
"""
```

#Missing Data

```python
#Finding Missing data
import numpy as np
import pandas as pd

nums = {
    "A":[1,2,np.nan,4,5],
    "B":[4,2,3,5,6],
    "C":[1,2,3,np.nan,np.nan],
    "D":[1,np.nan,np.nan,np.nan,5]
}
#np.nan= Null value

md=pd.DataFrame(nums)
md
```

```python
#Check is it null (isna), null=True, not null = False
md.isna()
```

```python
#Sum of null in a column
md.isna().sum()
```

```python
#is there any null in column
md.isna().any()
```

```python
md
```

```python
#remove all the rows that contain null
md.dropna()
```

```python
#keeps rows that have at least 3 non-NaN values.
md.dropna(thresh=3)
```

```python
#filling the missing data with same value
md.fillna("Hello") #temp
#md.fillna("Hello", inplace = True)#permanent
```

```python
#storing different values for different column
val = {
    "A":"Java",
    "B":"C#",
    "C":"Python",
    "D":"Js"
}
md.fillna(value=val)
```

#GroupByAggregation

#GroupBy

```python
data = {
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['s1','s1','s2','s2','s1','s2','s2','s1'],
    'Sales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01', periods =8)
}

gb = pd.DataFrame(data)
gb
```

```python
#Grpup by catagory A & B
cat = gb.groupby('Category')

for i, v in cat:
  print(i)
  print(v)

cat #group by create an object that can't see group A & B directly
```

```python
#sum of sales
cat = gb.groupby("Category")["Sales"].sum()
cat
```

```python
#group by store and calculate the sum of sales
cat = gb.groupby("Store")['Sales'].sum()
cat
```

```python
#Group by multiple column(category and store) and calculate the sum of sales
cat = gb.groupby(['Category',"Store"])['Sales'].sum()
cat
```

#Aggregation
mean, median, mode, min, max, count,std

```python
#calculates the average(mean)
gb['Sales'].mean()
```

```python
#can perform multiple calculation at once
gb["Sales"].agg(['sum','mean','min','max','count','std','median'])
```

```python
#multiple group to perform multiple calculation
gb.groupby('Category')['Sales'].agg(
    ['sum','mean','min','max','count']
)
```

#Pivot Tables

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

pt = pd.DataFrame(data)
pt
```

```python
pt.pivot_table(values = 'Sales', index = 'Region', columns = 'Product', aggfunc = 'median')
```

```python
pivot2 = pd.pivot_table(pt,values=['Sales','Units'], index = 'Region', columns = 'Product')
pivot2
```

#Cross Table

```python
pd.crosstab(pt['Region'],pt['Product'])
```
