```python
#if-elif
age = int (input ('Whats your age?'))

if age >= 18 :
  print("You are eligible to vote")

elif age < 18 :
  print("You are not eligible to vote")
```

```python
#if-elif-else
age = int (input("whats your age?"))

if (age < 18) :
  print("You are a kid")

elif (age >=18 and age <=60):
  print ("You are an adult")

else:
  print("You are a senior citizen")
```

```python
"""Take a traffic light code from user (Red,Yellow,Green) and give suggation
(Stop,Ready,Go) and if the user input not matching with light colours suggest
(Broken Light)"""

light = input("What's the light colour?")

if (light == "Red"):
  print('Stop')

elif(light =="Yellow"):
  print("Ready")

elif (light == "Green"):
  print("Go")

else:
  print("Broken Light")
```

```python
"""Take student marks as user input to give them grade-
if marks is more then or equal to 90 give -A
if marks is more then or equal to 80 give -B
if marks is more then or equal to 70 give -C
if marks is more then or equal to 50 give - D
if less then 50 give - Fail"""

marks = int (input("Enter your marks, i will tell your grade"))

if (marks >= 90):
  print("A")

elif (marks >= 80 and marks < 90):
  print("B")

elif (marks >=70 and marks < 80):
  print("C")

elif (marks >= 50 and marks < 70):
  print("D")

else:
  print("Fail")
```

```python
"""User can apply for driving license when they have NID and Birth_certificate and age range is not less then 16 and not more then 80.
1.if they have NID, Birth_certificate but age less then 15 then suggest "You are a kid you are not safe to drive"
2.if they have NID, Birth_certificate also age more then 18 and less then 70 suggest "You are an adult you are safe to drive"
3.if they have NID, Birth_certificate also age more then 70 Suggest "You are a senior citizen you are not safe to drive"
4.if they have Birth_certificate but not NID suggest "First Apply for NID"
5.Otherwise suggest "You can not apply for deiving licence"
"""
# Nested Conditions

NID = input("Do you have NID? (Yes/No): ")
Birth_certificate = input("Do you have Birth Certificate? (Yes/No): ")
age = int(input("Enter your age: "))

#1
if (NID == "Yes" and Birth_certificate == "Yes"):


    if (age < 16):
        print("You are a kid. You are not safe to drive.")


    elif (age >= 18 and age < 70):
        print("You are an adult. You are safe to drive.")


    elif (age > 70):
        print("You are a senior citizen. You are not safe to drive.")


    else:
        print("You are not eligible to apply for a driving licence.")

#2
elif (NID != "Yes" and Birth_certificate == "Yes"):
    print("First apply for NID.")

#3
elif (NID == "Yes" and Birth_certificate != "Yes"):
    print(" Need both NID and Birth certificate!")

#4
else:
    print("You cannot apply for a driving licence.")


```
