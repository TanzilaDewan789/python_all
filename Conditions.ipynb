```python
#if condition
age = 10
if (age < 18):
  print("You are Kid")
```

```python
#if-elif
age = 19
if (age < 18):
  print ("You are a kid")
elif (age >= 18):
  print("You are an adult")
```

```python
#if-elif-else
age = 60
if (age < 18):
  print ('You are a kid')

elif (age >=18 and age < 50):
  print("You are an adult")

else:
    print("You are a senior citizen")
```

```python
"""Take a traffic light code from user (Red,Green,Yellow) and give suggation
(Stop,Ready,Go) and if the user input not matching with light colours suggest
(Broken Light)"""

light = input("What is the colour of traffic light? ")

if (light == "Red"):
  print("Stop")

elif (light == "Green"):
  print('Go')

elif (light == " Yellow"):
  print("Ready")

else:
   print("Broken light")
```

```python
"""Take student marks as user input to give them grade-
if marks is more then or equal to 90 give -A
if marks is more then or equal to 80 give -B
if marks is more then or equal to 70 give -C
if marks is more then or equal to 50 give - D
if less then 50 give - Fail"""

marks = int (input ("What is your marks?") )

if(marks >= 90):
  print('Grade - A')
elif(marks >= 80 and marks <90):
  print("Grade -B")
elif(marks >=70 and marks <80):
  print("Grade -C")
elif(marks >=50 and marks <70):
  print("Grade -D")
else:
  print("Fail")
```
