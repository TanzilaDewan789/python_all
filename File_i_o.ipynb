#Mode-
1."r" → Read -open("file.txt", "r")

2."w" → Write

3."a" → Append -Adds data at end of file

4."x" → Exclusive create -Creates new file only and Gives error if file already exists

5."b" → Binary-Used for images, videos, files
Cannot use alone (combine with r/w/a) rb/wb/ab

6."+" → Read + Write (Allows both reading and writing)

open("file.txt", "r+")

open("file.txt", "w+")

open("file.txt", "a+")

#Automated File Sorter in File Explorer

```python
import os
import shutil

# Folder path(vs code or jupyter notebook)
path = r"C:/Users/name/Desktop/foldername/"

# Get all files
files = os.listdir(path)

# Create folders
if not os.path.exists(path + "csv files"):
    os.makedirs(path + "csv files")

if not os.path.exists(path + "image files"):
    os.makedirs(path + "image files")

if not os.path.exists(path + "txt files"):
    os.makedirs(path + "txt files")

# Move files
for file in files:

    if file.endswith(".csv"):
        shutil.move(path + file, path + "csv files/" + file)

    elif file.endswith(".png") or file.endswith(".jpg"):
        shutil.move(path + file, path + "image files/" + file)

    elif file.endswith(".txt"):
        shutil.move(path + file, path + "txt files/" + file)
```
