#File I/O -Input/ Output
Use to perform operation on a file (read, write...)

Types of file:

1.Text file: .txt, .docx, .log etc.

2.Binary file: .mp4, .mov, .png, .jpg etc

RAM(Random Access Memory)- volatile (Temporary) memory.Stores the code while it is running.

SSD(Solid State Drive) or HDD(Hard Disk Drive)-long-term storage & they store files, operating system, and programs.

#Mode-
1."r" → Read -open("file.txt", "r")

2."w" → Write

3."a" → Append -Adds data at end of file

4."x" → Exclusive create -Creates new file only and Gives error if file already exists

5."b" → Binary-Used for images, videos, files
Cannot use alone (combine with r/w/a) rb/wb/ab

6."+" → Read + Write (Allows both reading and writing)

open("file.txt", "r+")

open("file.txt", "w+")

open("file.txt", "a+")

#"w"- writing


```python
f = open("sample.txt", "w")
f.write("Do you want to learn Python?") #delete all the old data and overwrite new data
f.close()

f = open("sample.txt", "r")
print(f.read())
f.close()
```

#"a" - Append

```python
"""overwrite in txt file with numbers - 1,2,3,88,89,47,50,51,100
from that file containing numbers separated by comma, print the count of even numbers"""

#overwrite the file
with open("practice.txt","w") as f:
  f.write("1,2,3,88,89,47,50,51,100,90")

#read the file
with open("practice.txt","r") as f:
  data = f.read()
  print(data)
```

```python
count = 0

nums = data.split(",")
for val in nums:
  if (int(val)%2 == 0):
    count +=1

print(count)
```
