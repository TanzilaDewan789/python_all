#File I/O -Input/ Output
Use to perform operation on a file (read, write...)

Types of file:

1.Text file: .txt, .docx, .log etc.

2.Binary file: .mp4, .mov, .png, .jpg etc

RAM(Random Access Memory)- volatile (Temporary) memory.Stores the code while it is running.

SSD(Solid State Drive) or HDD(Hard Disk Drive)-long-term storage & they store files, operating system, and programs.


#Syntax

f = open("filename", "mode")

data = f.read()

print(data)

f.close()

```python
#creating a file- a(append) or w(write)
f = open("sample.txt", "a")
f.close()
```

```python
f = open("sample.txt","r")
data = f.read()
print(data)
print(type(data))
print(len(data))
f.close()
```

#Mode-
1."r" → Read -open("file.txt", "r")

2."w" → Write

3."a" → Append -Adds data at end of file

4."x" → Exclusive create -Creates new file only and Gives error if file already exists

5."b" → Binary-Used for images, videos, files
Cannot use alone (combine with r/w/a) rb/wb/ab

6."+" → Read + Write (Allows both reading and writing)

open("file.txt", "r+")

open("file.txt", "w+")

open("file.txt", "a+")

#1."r"- reading data

```python

f = open("sample.txt", "r") # Re-open the file in read mode
data = f.read(17)
print(data)
f.close() # Close the file after reading
```

```python
f = open("sample.txt", "r") # Re-open the file in read mode
line1 = f.readline()
print(line1)

line2 = f.readline()
print(line2)


f.close()
```

```python
f = open("sample.txt", 'r')
all_line = f.readlines()
print(all_line)
f.close()
```

#"w"- writing


```python
f = open("sample.txt", "w")
f.write("Do you want to learn Python?") #delete all the old data and overwrite new data
f.close()

f = open("sample.txt", "r")
print(f.read())
f.close()
```

#"a" - Append

```python

f = open("sample.txt", "a")
f.write("\nDo you know Pandas?")
f.close()
```

# "+" = reading + writing

```python
f = open("sample.txt", "r+")
f.write("ABC\n") #at the beginning of the file
f.close()
```
