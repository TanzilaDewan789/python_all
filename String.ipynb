#Concatenation of string
String concatenation means add strings together. Use the + character to add a variable to another variable.

```python
str1 = 'Muhammad ibn Musa al-Khwarizmiis,'
str2 ='considered the father of algorithms.'

print ('Concatenation of string :', str1+str2)

print ('Muhammad ibn Musa al-Khwarizmiis'+'considered the father of algorithms.' )
```

#Index of string:
Position of charecter, use to access character
1. Positive Index( Left to right) start from 0,1,2,3....9
2. Negative Index (Right to left) start from last word of the sentance. -10, -11, -9....1

```python
fruits = "Apple & Orange"

#what is 4th & last number character
print(fruits[3])#positive index start from 0
print(fruits[-1])#negative index start backword from -1
```

#Slicing
Accessing parts of string

```python
planet = "Venus_Earth"

print (planet[5])
print (planet[2:6])#Ending index dosen't include(Here E)

#[start, stop, step]
print(planet[2:7:2])

#reverse logic
print(planet[::-1])

#[first:5]
print(planet[0:5])
print(planet[:4])

#[5:last index]
print(planet[0:])

#[5:last index]
print(planet[5:])

#negative index
print(planet[-9:-5])
```

#String Function

```python
#print all the planets name in uppercase
planets ="mercury, venus, earth, mars, jupiter, saturn, uranus, neptune"
print(planets.upper())


#print all the planets name in lowercase
print(planets.lower())
```

```python
#Find if the last planets "saturn" or "neptune"?
print(planets.endswith('neptune'))
print(planets.endswith('saturn'))
```

```python
#Capitalized only 1st starting planet name
print(planets.capitalize())
```

```python
#Title
print(planets.title())
```

```python
#Replace 'earth' to 'our planet'
print(planets)

print(planets.replace('earth','our planet'))
```

```python
#Find u & count total number of u

print(planets.find('u'))
print(planets.count('u'))
```
