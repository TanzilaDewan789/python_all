#Tuples:
A built in data type that let us create immutable sequences of values.

```python
tup = (78,90,34,67,32,12,98,78,45,78)
print(tup)
print (type(tup))
#Access index

print(tup[3])
print(tup[-2])
```

```python
#Way to store tuple
t=()
print(type(t))
t1 = (1,)#only (1) is a int
print(type(t1))

name = ("John", 12)
print(type(name))
```

#Tuple Slicing

```python
tup = (78,90,34,67,32,12,98,78,45,78)
print(tup[2:7])
print(tup[:7])
print(tup[2:])
print(tup[-6:-2])

print(tup[::2])
```

#Methods

```python
tup = (78,90,34,67,32,12,98,78,45,78)

#.index(Values)- shows the index number
print(tup.index(98))

#.count(values)- Count total number of specific values
print(tup.count(78))
```

```python
#Sorted Asc
tup = (78,90,34,67,32,12,98,78,45,78)
num = sorted(tup)
print(num)

#Sorted Desc
tup = (78,90,34,67,32,12,98,78,45,78)
num2 = sorted(tup, reverse= True )
print(num2)
```
