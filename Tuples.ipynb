#Tuples:
A built in data type that let us create immutable sequences of values.

```python
tup = (78,90,34,67,32,12,98,78,45,78)
print(tup)
print(type(tup))

#Access index
print(tup[3])
print(tup[-3])
```

```python
#Way to store tuple
t = (1,)
print(t)
print(type(t))

t1 = (1)
print(t1)
print(type(t1))
```

#Tuple Slicing

```python
tup = (78,90,34,67,32,12,98,78,45,78)
print(tup[2:6])
print(tup[:6])
print(tup[2:])
print(tup[-8:-4])
```

#Methods

```python
tup = (78,90,34,67,32,12,98,78,45,78)

#Index(Values)- shows the indes number
print(tup.index(32))

#Count(values)- Count total number of specific values
print(tup.count(78))
```

```python
#Sorted Asc
tup = (78,90,34,67,32,12,98,78,45,78)
tup_sort = sorted(tup)
print(tup_sort)

#Sorted Desc
tup = (78,90,34,67,32,12,98,78,45,78)
tup_sort_desc = sorted(tup, reverse = True)
print(tup_sort_desc)
```
