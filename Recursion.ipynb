#Recursion
is a technique where function call itself repeatedly.

```python
#Print number n to 1
def show(n): #5,4,3,2,1
  if(n == 0):#base case
    return

  print(n)
  show(n-1)#recursive call
  print("end")#back track


show(5)

"""Visualization of the Call Stack || recursive call stack
show(5)
 └─ show(4)
     └─ show(3)
         └─ show(2)
             └─ show(1)
                 └─ show(0)

back track
show(1) -> print("end")
show(2) -> print("end")
show(3) -> print("end")
show(4) -> print("end")
show(5) -> print("end")"""
```

```python
#print factorial n = 5! = 5*4*3*2*1
"""How it work

Call stack:
f.fact(0)
e.fact(1)
d.fact(2)
c.fact(3)
b.fact(4)
a.fact(5)

=n*fact(n-1)

Back stack:
1 * 1 = 1
2 * 1 = 2
3 * 2 = 6
4 * 6 = 24
5 * 24 = 120"""

def fact(n):
  if(n==0 or n== 1):
    return 1

  else:
    return n * fact(n-1)


print(fact(5))
```

```python
#Calculate the sum of first n natural number
n = int(input("Enter n = "))

def num(n):
  if(n==0):
    return 0

  else:
    return n + num(n-1) # 5 + num(n-1)=4,3,2,1

print(num(n))
```

```python
#print all element in a list (parameter is lst and idx)
fruits = ["apple", "banana", "mango"]

def all_fruits(lst,idx):
  if(idx == len(lst)):
    return

  print(lst[idx])
  all_fruits(lst,idx+1)

print(all_fruits(fruits, 0))
```
