```python
#Print number n to 1
def show(n):
    if n == 0:
        return 0
    else:
        print(n)#call stack
        show(n - 1)#recursive call #if we return here it not going print back track
        print("end")#back track"""


show(5)

"""
call stack
5
4
3
2
1

recursive call stack
show(5)
 └─ show(4)
     └─ show(3)
         └─ show(2)
             └─ show(1)
                 └─ show(0)

back track
show(1) -> print("end")
show(2) -> print("end")
show(3) -> print("end")
show(4) -> print("end")
show(5) -> print("end")"""
```

```python
#print factorial n = 5! = 5*4*3*2*1
"""How it work

Call stack:
f.fact(0)
e.fact(1)
d.fact(2)
c.fact(3)
b.fact(4)
a.fact(5)

=n*fact(n-1)

Back stack:
1 * 1 = 1
2 * 1 = 2
3 * 2 = 6
4 * 6 = 24
5 * 24 = 120"""

def fact(n):
  if(n==0 or n== 1):
    return 1

  else:
    return n * fact(n-1)


print(fact(5))
```

```python
#Calculate the sum of first n natural number
n = int(input("Enter n = "))

def num(n):
  if(n==0):
    return 0

  else:
    return n + num(n-1) # 5 + num(n-1)=4,3,2,1

print(num(n))
```

```python
#print all element in a list (parameter is lst and idx)
fruits = ["apple", "banana", "mango"]

def all_fruits(lst,idx): #(fruits, 0 1 2)
  if(idx == len(lst)):#len(lst)=3
    return

  else:
    print(lst[idx]) #call stack
    all_fruits(lst,idx+1) #recursive call
    #no back track-(back track print reverse index)

print(all_fruits(fruits, 0))



"""call stack
all_fruits(fruits, 0)-apple
    ↓
all_fruits(fruits, 1)-banana
    ↓
all_fruits(fruits, 2)-mango
    ↓
all_fruits(fruits, 3)-None
    ↓
return

back track (nothing is print because there is no print statment after recursive call)
all_fruits(fruits, 2) returns
all_fruits(fruits, 1) returns
all_fruits(fruits, 0) returns"""
```

```python
#Fibinacci sequence -0, 1, 1, 2, 3, 5, 8, 13, 21,....
#idx-----------------0, 1, 2, 3, 4, 5, 6, 7, 8,.......

#print n= 4th of fibonacci

def fib(n):
  if(n==0):
    return 0
  elif(n==1):
    return 1
  else:
    return fib(n-1)+fib(n-2)

print(fib(4))

"""
fib(n)=fib(n-1)+fib(n-2)
fib(4)=fib(3)+fib(2)
      =fib(2)+fib(1)+fib(1)+fib(0)
      =fib(1)+fib(0)+1+1+0
      =1+0+1+1+0
      =3
"""
```

```python
#4. Print All Digits of Number
def digit(n):
  if( n < 10):
    print(n)#123
    return
  else:
    digit(n//10)
    print(n%10)

digit(123)

"""digit(123)
│
├── digit(12)
│   │
│   ├── digit(1)
│   │   └── print(1)
│   │
│   └── print(2)
│
└── print(3)"""
```

```python
#5. Binary to Decimal
def binary_to_decimal(binary):
    if binary < 10:
        return binary

    return binary_to_decimal(binary // 10) * 2 + (binary % 10)

print(binary_to_decimal(10100))


"""
binary_to_decimal(10100)
│
├─ binary_to_decimal(1010)
│  │
│  ├─ binary_to_decimal(101)
│  │  │
│  │  ├─ binary_to_decimal(10)
│  │  │  │
│  │  │  ├─ binary_to_decimal(1)
│  │  │  │   └─ 1
│  │  │  │
│  │  │  └─ 1×2+0 = 2
│  │  │
│  │  └─ 2×2+1 = 5
│  │
│  └─ 5×2+0 = 10
│
└─ 10×2+0 = 20"""
```
