```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Type of chart -

| **Plot** | **Use when you want to show** | **Memory trick** |
|---|---|---|
| **Line plot**(Time vs Numerical) | Trend or change over time | Time → Line |
| **Scatter plot**(Numerical vs Numerical) | Relationship between two numerical variables | Relationship → Scatter |
| **Histogram**(Numerical distribution) | Distribution of a numerical variable | Distribution → Hist |
| **Bar plot**(Categorical vs Numerical)/(Categorical frequency) | Compare values across categories | Comparison → Bar |
| **Pie chart**(Categorical proportion or percentage)| Proportions or percentages of a whole | Share → Pie |

#2D line plot -
use to show how a value changes over time.

those known as time series data.

example: tmeperature, how temperature(y axis) change on time(x axis- days, week, months or years).

```python
hour =[2,3,4,1,5,7,3]
days =[1,2,3,4,5,6,7]

plt.plot(days,hour)#(x,y)
```

```python
plt.xlabel('Days')
plt.ylabel('Study of hours')
plt.title('Trend of a student study time over a week')

plt.plot(days,hour)
```

```python
#Multiple data ploting

hour_1 =[2,3,4,1,5,7,3]#student 1
hour_2 =[1,4,3,5,7,2,1]#student 2

days =[1,2,3,4,5,6,7]

plt.xlabel("Days")
plt.ylabel("Study hours")
plt.title("Trend of 2 students study time over a week")

plt.plot(days, hour_1)
plt.plot(days, hour_2)

plt.show()
```

#Customizing line plot with color, linestyle, marker

```python
#multiple ploting on df
plt.xlabel('Year')
plt.ylabel('Count of student enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Programming'],
         label = 'Programming',
         color = '#cc2b83',
         linewidth = 3,
         marker ='o',#show the intercet point of x and y
         markersize =7)

plt.plot(df['Year'], df['Digital Marketing'],
         label = 'Digital Marketing',
         color = 'black',
         linestyle = 'dotted')


plt.legend(loc = 'best')
plt.grid()#show the grid line like graph paper

plt.show()
```

#Scatter plot
To find correlation between freatures(columns)

3 types of correlation -

+1 = positive correlation

-1 = negative correlation

0 = no relation

can be use to-
1. identify outliers and how its effect
2. to select feature columns
3. to understand correlation to choose ML algorithm
4. to see the performance of the model after train


```python
students = pd.read_csv('/content/student_IQdata.csv')
students
```

```python
#ralation on study_hours and IQ_score (like possitive correlation - linear line)

plt.xlabel('Study hours')
plt.ylabel('IQ score')
plt.title('IQ score based on Study hours \n Possitive correlation')

plt.scatter( students['Study_Hour'], students['IQ_Score'])

#outliears - 'Study_Hour' =2.5 and  'IQ_Score' = 150 (we have to skip outliers during model taining)
```

```python
# relation on Shoe_Size and IQ_score (No correlation)

plt.xlabel('Shoe_Size')
plt.ylabel('IQ score')
plt.title('Relation between shoe size and IQ score \n No correlation')

plt.scatter( students['Shoe_Size'], students['IQ_Score'])

#we can remove or skip the shoe_size column during model training
```

```python
#ralation between 'Chilling_Hours' & 'IQ_Score' (Negative correlations)


plt.xlabel('Chilling_Hours')
plt.ylabel('IQ score')
plt.title('Relation between shoe size and IQ score \n Negative correlation')

plt.scatter( students['Chilling_Hours'], students['IQ_Score'],
            color = 'purple',
             edgecolors= 'black')

```

#Histogram
use for
1. how data distributed over a factor
2. frequency of a data
3. check outliers

Type of distributions-
1. Lef-skewed distribution(Negative skew)=> (left side less frequency)
2. Right-skewed distribution(Positive skew)=> (right side less frequency)
3. Symmetrical distribution / bell curve ( left & right both contain equal frequency) - give less bias prediction


```python
plt.hist( students['IQ_Score'], color = 'pink', edgecolor ="black")
plt.xlabel('IQ score')
plt.ylabel('Frequency of students')
plt.title('Right-skewed')
```

```python
plt.hist( students['Study_Hour'],
         color = 'lightblue',
          edgecolor ="black",
          bins = [1,2,3,4,5,6,7,8])

plt.xlabel('Study hours')
plt.ylabel('Frequency of students')
plt.title('Kind of fairly distributed')
```

#Bar Chart

use on-
Catagorical values

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
#gender wise catagorize

gender_group = df.groupby('Gender').size()
gender_group.index #For x axis
#gender_group.values #For y axis
```

```python
plt.bar(gender_group.index,
        gender_group.values,
         color = ['lightpink','lightblue'],
        edgecolor = 'black')

plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender wise catagorize')
```

```python
#bar chart on real data

df = pd.read_csv('/content/student_data (1).csv')
df.head()
```

```python
status = df.groupby('CompletionStatus').size()
status.index
```

```python
plt.bar(status.index,
        status.values,
        color = ['lightgreen', 'lightblue','red'],
        edgecolor = 'black')

plt.xlabel('CompletionStatus')
plt.ylabel('Count')
plt.title("student's CompletionStatus")
```

```python
location = df.groupby('Location').size()
location
```

```python
plt.bar(location.index,
        location.values,
        color = ['purple','lightgreen','lightpink','lightblue','lightyellow'],
        edgecolor = 'black')

plt.xlabel('Location')
plt.ylabel('Count')
plt.title('Students Location')
```

#Pie chart
give graphical representation on percentage

```python
status = df.groupby('CompletionStatus').size()
status
```

```python
plt.pie(status,
        labels= status.index,
        autopct= '%1.1f%%',
        explode = (0.1, 0, 0),
        shadow= True)

plt.show()
```
