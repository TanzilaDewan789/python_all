```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#2D line plot -
use to show how a value changes over time.

those known as time series data.

example: tmeperature, how temperature(y axis) change on time(x axis- days, week, months or years).

```python
hour =[2,3,4,1,5,7,3]
days =[1,2,3,4,5,6,7]

plt.plot(days,hour)#(x,y)
```

```python
plt.xlabel('Days')
plt.ylabel('Study of hours')
plt.title('Trend of a student study time over a week')

plt.plot(days,hour)
```

```python
#Multiple data ploting

hour_1 =[2,3,4,1,5,7,3]#student 1
hour_2 =[1,4,3,5,7,2,1]#student 2

days =[1,2,3,4,5,6,7]

plt.xlabel("Days")
plt.ylabel("Study hours")
plt.title("Trend of 2 students study time over a week")

plt.plot(days, hour_1)
plt.plot(days, hour_2)

plt.show()
```

#2D line plot from Pandas

```python
df = pd.read_csv('/content/enrollment_data.csv')
df
```

```python
plt.xlabel('Year')
plt.ylabel('Count of enrollment')
plt.title('Students enrollment over 8 years')

plt.plot(df['Year'], df['Programming'])# x = year, y = programming
```

```python
#multiple ploting on df
plt.xlabel('Year')
plt.ylabel('Count of student enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Programming'], label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital Marketing')


plt.legend(loc = 'best')#its show lebels name- can avoid your loc
```

#Customizing line plot with color, linestyle, marker

```python
#multiple ploting on df
plt.xlabel('Year')
plt.ylabel('Count of student enrollment')
plt.title('Student enrollment over 8 years')

plt.plot(df['Year'], df['Programming'],
         label = 'Programming',
         color = '#cc2b83',
         linewidth = 3,
         marker ='o',#show the intercet point of x and y
         markersize =7)

plt.plot(df['Year'], df['Digital Marketing'],
         label = 'Digital Marketing',
         color = 'black',
         linestyle = 'dotted')


plt.legend(loc = 'best')
plt.grid()#show the grid line like graph paper

plt.show()
```

#Scatter plot
To find correlation between freatures(columns)

3 types of correlation -

+1 = positive correlation

-1 = negative correlation

0 = no relation

can be use to-
1. identify outliers and how its effect
2. to select feature columns
3. to understand correlation to choose ML algorithm
4. to see the performance of the model after train


```python
students = pd.read_csv('/content/student_IQdata.csv')
students
```

```python
#ralation on study_hours and IQ_score (like possitive correlation - linear line)

plt.xlabel('Study hours')
plt.ylabel('IQ score')
plt.title('IQ score based on Study hours \n Possitive correlation')

plt.scatter( students['Study_Hour'], students['IQ_Score'])

#outliears - 'Study_Hour' =2.5 and  'IQ_Score' = 150 (we have to skip outliers during model taining)
```

```python
# relation on Shoe_Size and IQ_score (No correlation)

plt.xlabel('Shoe_Size')
plt.ylabel('IQ score')
plt.title('Relation between shoe size and IQ score \n No correlation')

plt.scatter( students['Shoe_Size'], students['IQ_Score'])

#we can remove or skip the shoe_size column during model training
```

```python
#ralation between 'Chilling_Hours' & 'IQ_Score' (Negative correlations)


plt.xlabel('Chilling_Hours')
plt.ylabel('IQ score')
plt.title('Relation between shoe size and IQ score \n Negative correlation')

plt.scatter( students['Chilling_Hours'], students['IQ_Score'],
            color = 'purple',
             edgecolors= 'black')

```
