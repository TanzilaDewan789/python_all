#List Slicing


```python
id = [101,105,103,108,106,67,89,35,78]
print(id)

#positive
print(id[1:4])
print(id[2:])
print(id[:5])

#negative
print(id [-6:-1])
print(id [:-1])
print(id [-6:])

#rev
print(id[::-1])#start, stop, step(-1 means negative)
```

#List Method

```python
#Append- Add one element at the end of list
student2 = ['Bob', 25, 3.4, "CSE"]
print(student2)

#.append(value)- Add country as one element at the end of list
student2.append("Dhaka")
print(student2)

#.insert(index, value)- insert into specific index
student2.insert(1, "Alex")
print(student2)

#replace
student2[3] = 3.8
print(student2)


#.remove(value)- delete any value
student2.remove("Alex")
print(student2)

#.pop(index num) - delete any index
student2.pop(1)
student2

#.reverse()- reverses in backword
student2.reverse()
student2
```

```python

course_id= [23,2,56,90,56,89,23,1]
print(course_id)

#Sort() - Ascending order
course_id.sort()
print(course_id)


#descending order
course_id.sort(reverse=True)
print(course_id)
```

```python
num =[ 1,2,3,4,5]
print(num)

#Copy()- make a shallow copy of list
num.copy()
print(num)

#reverse
num.reverse()
print(num)
#Clear()- delete all values
num.clear()
print(num)
```
