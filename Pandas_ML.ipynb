#Loops

df.iterrows() → Iterates through each row.

i → Stores the row index.


series → Stores the row data as a Pandas Series.

```python
for i,series in df.iterrows():
  print(f'{i} : {series}')

#i = Row index
#series = Entire row
```

```python
for i in df.itertuples(index = False):
  print(i)
```

#Filtering data based on condition

```python
completed = df.loc[df['CompletionStatus']=="Completed"]
completed
```

```python
#completed and ds marks is greater or equal 90

completed_ds90 =df.loc[(df['CompletionStatus']=="Completed") & (df['Data Structure Marks']>=90)]
completed_ds90
```
