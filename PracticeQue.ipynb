```python
#Q12. Create a number guessing game.
import random

guess = int(input("Enter any number from 0 to 10: "))
num = random.randint(0,5)

if (guess == num):
  print(f"correct! You guess the correct number{num} ")

elif(guess>num):
  print("Too high")

elif(guess<num):
  print("Too low")
```

```python
#check a number is prime or not
#A prime number is divisible only by 1 and itself.(7/1 and 7/7 )- 1 and 0 are not prime number.

num = int(input())

if (num==0 or num==1):
  print("not prime")

else:
  for i in range(2,num):
    if(num%i==0):
      print("Not prime")
      break
  else:
    print("prime")
```

```python
#Q14. Check if a given year is a leap year or not.
#It is divisible by 4 but not divisible by 100 also divisible by 400 is a leap year
year = int(input("Enter any Year: "))
if((year%4 == 0 and year%100 != 0)or year % 400 ==0):
  print("Leap Year")
else:
  print("Not a leap year")
```

```python
#Q16. Write a program to find the GCD of two numbers.
a = int(input())
b = int(input())

#divisor of a
a_store =[]
for i in range(1,a+1):
  if(a%i==0):
    a_store.append(i)

#divisor of b
b_store =[]
for j in range(1,b+1):
  if(b%j==0):
    b_store.append(j)

#common
common=[]

for k in a_store:
  if k in b_store:
    common.append(k)

print(max(common))



```
