```python
#Q12. Create a number guessing game.
import random

target = random.randint(1,10)

while True:
  user_choice= input("Guess the terget or quit(Q): ")
  if(user_choice == "Q"):
    break

  user_choice = int(user_choice)
  if (user_choice== target):
    print("Success: Correct Guess!")
    break
  elif(user_choice < target):
    print("Hint: Guess Higher")
  else:
    print("Hint: Guess Lower")

print("_____GAME OVER_______")
```

```python
#check a number is prime or not
#A prime number is divisible only by 1 and itself.(7/1 and 7/7 )- 1 and 0 are not prime number.

num = int(input())

if (num==0 or num==1):
  print("not prime")

else:
  for i in range(2,num):
    if(num%i==0):
      print("Not prime")
      break
  else:
    print("prime")
```

```python
#Q16. Write a program to find the GCD of two numbers.(greatest common divisor (GCD))
a = int(input())
b = int(input())

#divisor of a
a_store =[]
for i in range(1,a+1):
  if(a%i==0):
    a_store.append(i)

#divisor of b
b_store =[]
for j in range(1,b+1):
  if(b%j==0):
    b_store.append(j)

#common
common=[]

for k in a_store:
  if k in b_store:
    common.append(k)

print(max(common))



```

```python
#Q17. Write a program to find the LCM of two numbers.(the smallest positive integer that is a multiple of two or more numbers)
a = 4
b = 6

lcm =max(a,b)

for i in range(lcm,a*b+1):
  if(i%a == 0 and i%b ==0):
    print(i)
    break
```

```python
#Q19. Write a program to calculate the sum of digits of a number.
num = 342
total =0

while num>0:
  digit = num%10 #34(2)->3(4)->(3)
  total = total + digit #0+2=2+4=6+3=9

  num = num//10 ##342//10=34//10=3//10=0 [loop run for 3 times until num = 0 for while condition]

print(total)
```

```python
#Q20. Create a program to find the second largest number in a list.
num = [22,22,12,5,50,4,3,50]
largest = max(num)

while largest in num:
  num.remove(largest)

print(max(num))
```
