```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
student = pd.read_csv('/content/sns_data (1).csv')
student.head()
```

# Lineplot using Seaborn

```python
#axis level
sns.lineplot(data = student, x ='week', y ='attendance_rate', errorbar=None)
```

```python
sns.lineplot(data = student,
             x ='week',
             y = 'attendance_rate',
             errorbar = 'se',
             hue= 'gender')# Creates separate colored lines for each gender.
```

```python
#figure level
#relplot()-Relational Plot=> support 2 types of plot
#kind='scatter' → Scatter plot & kind='line' → Line plot
sns.relplot(data = student,
            x = 'week',
            y = 'attendance_rate',
            errorbar = None,
            kind = 'line')
```

```python
sns.relplot(kind='line',
            data = student,
            x = 'week',
            y = 'attendance_rate',
            errorbar = None,
            hue = 'gender')
```

```python
sns.relplot(kind= 'line',
            data = student,
            x = 'week',
            y= 'test_score',
            errorbar = None,
            hue='class_level'
            )
```

```python
# Scatter plot using Seaborn
sns.scatterplot(data = student,
                x = 'study_hours',
                y = 'test_score',
                hue='gender',
                style='subject',
                size = 'Tshirt_size'
                )
```

```python
sns.relplot(
           data= student,
           x = 'study_hours',
           y = 'test_score',
           hue= 'gender',
           style = 'subject',
           size = 'Tshirt_size',
           kind= 'scatter')
```

```python
tips_data = sns.load_dataset('tips')#Seaborn's built-in Tips dataset
tips_data.head()
```

```python
sns.relplot(kind = 'scatter',
            data = tips_data,
            x='total_bill',
            y = 'tip',
            hue= 'sex',#Changes the color based on categories (groups).
            style ='time',#Changes the marker shape based on the time category.
            size='size')#Changes the marker size based on the number of people (size column)
```

```python
sns.scatterplot(data = tips_data,
                x = 'total_bill',
                y = 'tip',
                hue= 'sex',
                style = 'time',
                size = 'size')
```

```python
student = pd.read_csv('/content/student_dataset_complete.csv')
student.head()
```

#Facet Plot

```python
sns.scatterplot(data = student,
                x= 'study_hours',
                y = 'test_score',
                hue= 'gender')
```

```python
sns.relplot(data =student,
            x = 'study_hours',
            y = 'test_score',
            hue= 'gender',
            kind = 'scatter')
```

```python
sns.relplot(data= student,
            x = 'study_hours',
            y = 'test_score',
            col= 'gender')#creates separate subplots (columns) based on categories.
```

```python
sns.relplot(data = student,
            x= 'study_hours',
            y = 'test_score',
            col= 'gender',
            row= 'hostel',
            kind = 'scatter')
```

```python
sns.relplot(data = student,
            x = 'study_hours',
            y = 'test_score',
            col= 'week',
            col_wrap = 2,#Displays a maximum of 2 plots per row
            kind= 'scatter')
```
