# Feature Extraction

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'/content/anime.csv')
```

```python
df.head()
```

```python
#title
df.loc[2]['Title']
```

```python
#make a new column for episode count
def extract_episodes(txt):
    data = ""
    check = False

    for i in txt:

        if i == "(":
            check = True
            continue      # skip '('

        if i == ")":
            break

        if check:
            data += i

    return data
```

```python
#new columns for episode
df['Episodes'] =df['Title'].apply(extract_episodes)
```

```python
#removing eps after int
df['Episodes']=df['Episodes'].str.replace(" eps","")
```

```python
#changing datatype str to int
df['Episodes']= df['Episodes'].astype(int)
```

```python
#make a new column for the tile stamp
df.loc[0]['Title']
```

```python
def extract_timestamp(txt):

    months = ["Jan","Feb","Mar","Apr","May","Jun",
              "Jul","Aug","Sep","Oct","Nov","Dec"]

    start = len(txt)

    # Find the earliest month in the text
    for month in months:
        pos = txt.find(month)

        if pos != -1 and pos < start:
            start = pos

    year_count = 0

    for i in range(start, len(txt)):

        if txt[i:i+4].isdigit():

            year_count += 1

            if year_count == 2:
                return txt[start:i+4]
```

```python
df["Total Time"] = df["Title"].apply(extract_timestamp)
```

```python
df
```

```python
month_dict = {
    "Jan":1, "Feb":2, "Mar":3, "Apr":4,
    "May":5, "Jun":6, "Jul":7, "Aug":8,
    "Sep":9, "Oct":10, "Nov":11, "Dec":12
}

def total_months(txt):

    start_month = month_dict[txt[:3]]
    start_year = int(txt[4:8])

    end_month = month_dict[txt[11:14]]
    end_year = int(txt[15:19])

    return (end_year - start_year) * 12 + (end_month - start_month) + 1

# Apply the function AFTER it is defined
df["Months"] = df["Total Time"].apply(total_months)
```
